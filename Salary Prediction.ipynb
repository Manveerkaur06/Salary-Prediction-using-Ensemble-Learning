## 4. Separate Input and Output

```python
X = data[['YearsExperience']]
y = data['Salary']
X.head(), y.head()
```

## 5. Training and Testing Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=0
)
print('Training rows:', len(X_train))
print('Testing rows:', len(X_test))
```

## 6. Train the Models

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

tree_model = DecisionTreeRegressor(max_depth=3, random_state=0)
tree_model.fit(X_train, y_train)

forest_model = RandomForestRegressor(
    n_estimators=100, max_depth=4, random_state=0
)
forest_model.fit(X_train, y_train)
```

## 7. Create the Ensemble Prediction

The final prediction is the simple average of Linear Regression and Random Forest predictions.

```python
linear_prediction = linear_model.predict(X_test)
tree_prediction = tree_model.predict(X_test)
forest_prediction = forest_model.predict(X_test)
ensemble_prediction = (linear_prediction + forest_prediction) / 2
```

## 8. Evaluate the Models

```python
def calculate_metrics(actual, predicted):
    mse = mean_squared_error(actual, predicted)
    return {
        'MAE': mean_absolute_error(actual, predicted),
        'MSE': mse,
        'RMSE': np.sqrt(mse),
        'R2 Score': r2_score(actual, predicted),
    }

predictions = {
    'Linear Regression': linear_prediction,
    'Decision Tree': tree_prediction,
    'Random Forest': forest_prediction,
    'Final Ensemble': ensemble_prediction,
}

metrics = pd.DataFrame({
    name: calculate_metrics(y_test, values)
    for name, values in predictions.items()
}).T
metrics.round(4)
```

```python
plt.figure(figsize=(8, 5))
axis = sns.barplot(x=metrics.index, y=metrics['R2 Score'], color='#3b8791')
axis.set_ylim(0, 1.05)
axis.set_title('Model Comparison Using R2 Score')
axis.tick_params(axis='x', rotation=12)
plt.show()
```

## 9. Make a Sample Prediction

```python
plt.figure(figsize=(7, 6))
plt.scatter(y_test, ensemble_prediction, color='#2b6f77', s=75)
lower = min(y_test.min(), ensemble_prediction.min())
upper = max(y_test.max(), ensemble_prediction.max())
plt.plot([lower, upper], [lower, upper], '--', color='#e07a5f',
         label='Ideal prediction line')
plt.title('Actual Salary vs Ensemble Prediction')
plt.xlabel('Actual Salary')
plt.ylabel('Predicted Salary')
plt.legend()
plt.show()
```

```python
sample = pd.DataFrame({'YearsExperience': [5.0]})
linear_value = linear_model.predict(sample)[0]
forest_value = forest_model.predict(sample)[0]
final_value = (linear_value + forest_value) / 2
print(f'Linear Regression: ₹{linear_value:,.0f}')
print(f'Random Forest: ₹{forest_value:,.0f}')
print(f'Final Ensemble Salary: ₹{final_value:,.0f}')
```
